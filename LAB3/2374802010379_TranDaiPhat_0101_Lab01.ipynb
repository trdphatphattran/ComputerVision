## 3. Show hình ảnh ra cho trực quan, sử dụng plt.imshow. Có nhận xét gì về bức ảnh? Có cách nào để hiển thị đúng hay không?

Cách 1, đảo chiều matrix; cách 2, convert các chế độ màu.

### Nhận xét:  
OpenCV đọc ảnh theo BGR, nên cần đổi sang RGB khi dùng với matplotlib

```python
# Cách 1 đảo chiều matrix
import cv2
import numpy as np
import matplotlib.pyplot as TranDaiPhat
from google.colab.patches import cv2_imshow

# đọc ảnh
image = cv2.imread("/content/drive/MyDrive/picture1.png")
# đảo ảnh trái <-> phải
flip_horizontal = cv2.flip(image, 1)
# đảo ảnh trên <-> dưới
flip_vertical = cv2.flip(image, 0)

TranDaiPhat.subplot(1,3,1)
TranDaiPhat.imshow(flip_horizontal)

TranDaiPhat.subplot(1,3,2)
TranDaiPhat.imshow(flip_vertical)

TranDaiPhat.show()
```

```python
# Cách 2 convert các chế độ màu
# Cách 2: Convert chế độ màu (BGR -> RGB)
import cv2
import matplotlib.pyplot as plt

# đọc ảnh (OpenCV đọc ở BGR)
image = cv2.imread("/content/drive/MyDrive/picture1.png")

# convert BGR -> RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

TranDaiPhat.subplot(1, 2, 1)
TranDaiPhat.imshow(image)

TranDaiPhat.subplot(1, 2, 2)
TranDaiPhat.imshow(image_rgb)

TranDaiPhat.show()
```

## 5. Nếu muốn crop ảnh thì chúng ta làm gì?

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# đọc ảnh
image = cv2.imread("/content/drive/MyDrive/picture1.png")

left  = 100
upper = 50
right = 400
lower = 300

img_crop = image[upper:lower, left:right]
TranDaiPhat.imsave("hinh2.png", img_crop)
TranDaiPhat.imshow(img_crop)
```

## 6. Ghép dọc và ghép ngang 2 ảnh

```python
# ghép ngang 2 ảnh
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img1 = cv2.imread("hinh1.png")
img2 = cv2.imread("hinh2.png")

h = min(img1.shape[0], img2.shape[0])
img1 = cv2.resize(img1, (int(img1.shape[1] * h / img1.shape[0]), h))
img2 = cv2.resize(img2, (int(img2.shape[1] * h / img2.shape[0]), h))

result = np.hstack((img1, img2))

TranDaiPhat.imshow(result)


```

```python
# ghép dọc 2 ảnh
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img1 = cv2.imread("hinh1.png")
img2 = cv2.imread("hinh2.png")

w = min(img1.shape[1], img2.shape[1])
img1 = cv2.resize(img1, (w, int(img1.shape[0] * w / img1.shape[1])))
img2 = cv2.resize(img2, (w, int(img2.shape[0] * w / img2.shape[1])))

result = np.vstack((img1, img2))

TranDaiPhat.imshow(result)
```
